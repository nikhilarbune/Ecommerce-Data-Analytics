```python
import pandas as pd
import os

# Raw data folder
data_path = "../data/raw"

# Get all CSV files
files = [file for file in os.listdir(data_path) if file.endswith(".csv")]

# Load all CSV files
dataframes = {}

for file in files:
    file_path = os.path.join(data_path, file)

    df = pd.read_csv(file_path)

    dataframes[file] = df

    print("=" * 70)
    print(f"File: {file}")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
    print("Columns:", list(df.columns))
    
```

```python
# Check missing values and duplicates

for file, df in dataframes.items():
    print("=" * 70)
    print(f"File: {file}")
    
    print("\nMissing Values:")
    print(df.isnull().sum())
    
    print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Missing value percentage for each dataset

for file, df in dataframes.items():
    missing = df.isnull().sum()
    missing_percent = (missing / len(df)) * 100
    
    result = pd.DataFrame({
        "Missing Values": missing,
        "Missing Percentage": missing_percent.round(2)
    })
    
    result = result[result["Missing Values"] > 0]
    
    print("=" * 70)
    print(f"File: {file}")
    
    if result.empty:
        print("No missing values found.")
    else:
        print(result)
```

```python
# Check data types of all datasets

for file, df in dataframes.items():
    print("=" * 70)
    print(f"File: {file}")
    print(df.dtypes)
```

```python
# Create copies for cleaning

cleaned_data = {}

for file, df in dataframes.items():
    cleaned_data[file] = df.copy()

print("Cleaned dataset copies created successfully.")
```

```python
# Convert date columns to datetime

date_columns = {
    "olist_orders_dataset.csv": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    
    "olist_order_items_dataset.csv": [
        "shipping_limit_date"
    ],
    
    "olist_order_reviews_dataset.csv": [
        "review_creation_date",
        "review_answer_timestamp"
    ]
}

for file, columns in date_columns.items():
    for column in columns:
        cleaned_data[file][column] = pd.to_datetime(
            cleaned_data[file][column],
            errors="coerce"
        )

print("Date columns converted successfully.")
```

```python
# Load order items and cleaned products

order_items = cleaned_data[
    "olist_order_items_dataset.csv"
].copy()

products_cleaned = cleaned_data[
    "products_cleaned"
].copy()

print("Order Items:", order_items.shape)
print("Products:", products_cleaned.shape)
```

```python
# Merge order items with product information

sales_items = order_items.merge(
    products_cleaned[
        [
            "product_id",
            "product_category",
            "product_weight_g",
            "product_length_cm",
            "product_height_cm",
            "product_width_cm"
        ]
    ],
    on="product_id",
    how="left"
)

print("Sales Items after merge:", sales_items.shape)

print("\nSample:")
print(
    sales_items[
        [
            "order_id",
            "product_id",
            "product_category",
            "price",
            "freight_value"
        ]
    ].head(10)
)
```

```python
# Check sales data quality

print("Missing values:")
print(
    sales_items[
        ["price", "freight_value", "product_category"]
    ].isnull().sum()
)

print("\nInvalid price records:", (sales_items["price"] <= 0).sum())
print(
    "Invalid freight records:",
    (sales_items["freight_value"] < 0).sum()
)
```

```python
# Calculate sales metrics

sales_items["revenue"] = sales_items["price"]

sales_items["total_order_value"] = (
    sales_items["price"] +
    sales_items["freight_value"]
)

print(
    sales_items[
        [
            "order_id",
            "price",
            "freight_value",
            "revenue",
            "total_order_value"
        ]
    ].head(10)
)
```

```python
# Overall Sales KPIs

total_revenue = sales_items["revenue"].sum()

total_freight = sales_items["freight_value"].sum()

total_sales_value = sales_items["total_order_value"].sum()

total_items = len(sales_items)

unique_orders = sales_items["order_id"].nunique()

average_order_value = (
    sales_items.groupby("order_id")["total_order_value"]
    .sum()
    .mean()
)

print("===== SALES KPIs =====")
print(f"Total Revenue: R$ {total_revenue:,.2f}")
print(f"Total Freight: R$ {total_freight:,.2f}")
print(f"Total Sales Value: R$ {total_sales_value:,.2f}")
print(f"Total Items Sold: {total_items:,}")
print(f"Unique Orders: {unique_orders:,}")
print(f"Average Order Value: R$ {average_order_value:,.2f}")
```

```python
# Category-wise sales analysis

category_sales = (
    sales_items
    .groupby("product_category")
    .agg(
        revenue=("revenue", "sum"),
        freight=("freight_value", "sum"),
        items_sold=("product_id", "count"),
        average_price=("price", "mean")
    )
    .sort_values("revenue", ascending=False)
)

print(category_sales.head(15))
```

```python
print("\nTop 10 Revenue Categories:")
print(category_sales.head(10))
```

```python
# Merge sales with order purchase dates

sales_analysis = sales_items.merge(
    orders[
        [
            "order_id",
            "order_purchase_timestamp",
            "order_status"
        ]
    ],
    on="order_id",
    how="left"
)

print("Sales analysis shape:", sales_analysis.shape)

print(
    sales_analysis[
        [
            "order_id",
            "order_purchase_timestamp",
            "order_status",
            "product_category",
            "revenue"
        ]
    ].head()
)
```

```python
# Monthly revenue trend

sales_analysis["order_month"] = (
    sales_analysis["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_sales = (
    sales_analysis
    .groupby("order_month")
    .agg(
        revenue=("revenue", "sum"),
        orders=("order_id", "nunique"),
        items=("product_id", "count")
    )
    .reset_index()
)

print(monthly_sales)
```

```python
# Monthly revenue for delivered orders only

delivered_sales = sales_analysis[
    sales_analysis["order_status"] == "delivered"
].copy()

delivered_sales["order_month"] = (
    delivered_sales["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_delivered_sales = (
    delivered_sales
    .groupby("order_month")
    .agg(
        revenue=("revenue", "sum"),
        orders=("order_id", "nunique"),
        items=("product_id", "count")
    )
    .reset_index()
)

print(monthly_delivered_sales)
```

```python
# Best performing month

best_month = monthly_delivered_sales.loc[
    monthly_delivered_sales["revenue"].idxmax()
]

print("Best Revenue Month:")
print(best_month)
```

```python
# Month-over-month revenue growth

monthly_delivered_sales["revenue_growth_%"] = (
    monthly_delivered_sales["revenue"]
    .pct_change()
    .mul(100)
    .round(2)
)

print(monthly_delivered_sales)
```

```python
# Customer analysis dataset

customers = cleaned_data[
    "olist_customers_dataset.csv"
].copy()

customer_orders = orders[
    [
        "order_id",
        "customer_id",
        "order_purchase_timestamp",
        "order_status"
    ]
].copy()

customer_sales = sales_items[
    [
        "order_id",
        "revenue",
        "freight_value"
    ]
].copy()

customer_analysis = customer_orders.merge(
    customers[
        [
            "customer_id",
            "customer_unique_id",
            "customer_city",
            "customer_state"
        ]
    ],
    on="customer_id",
    how="left"
)

customer_analysis = customer_analysis.merge(
    customer_sales.groupby("order_id", as_index=False)
    .agg(
        revenue=("revenue", "sum"),
        freight=("freight_value", "sum")
    ),
    on="order_id",
    how="left"
)

print("Customer Analysis Shape:", customer_analysis.shape)

print(
    customer_analysis[
        [
            "order_id",
            "customer_unique_id",
            "customer_city",
            "customer_state",
            "revenue"
        ]
    ].head(10)
)
```

```python
# Customer purchase frequency

customer_frequency = (
    customer_analysis
    .groupby("customer_unique_id")
    .agg(
        orders=("order_id", "nunique"),
        revenue=("revenue", "sum")
    )
    .reset_index()
)

customer_frequency["customer_type"] = customer_frequency[
    "orders"
].apply(
    lambda x: "Repeat Customer" if x > 1 else "One-Time Customer"
)

print(customer_frequency["customer_type"].value_counts())

print("\nCustomer Type Percentage:")
print(
    customer_frequency["customer_type"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Revenue contribution by customer type

customer_type_revenue = (
    customer_frequency
    .groupby("customer_type")
    .agg(
        customers=("customer_unique_id", "count"),
        total_revenue=("revenue", "sum"),
        average_revenue=("revenue", "mean")
    )
    .reset_index()
)

customer_type_revenue["revenue_percentage"] = (
    customer_type_revenue["total_revenue"]
    / customer_type_revenue["total_revenue"].sum()
    * 100
).round(2)

print(customer_type_revenue)
```

```python
# Top 10 customers by revenue

top_customers = (
    customer_frequency
    .sort_values("revenue", ascending=False)
    .head(10)
)

print(top_customers)
```

```python
# State-wise customer analysis

state_analysis = (
    customer_analysis
    .groupby("customer_state")
    .agg(
        customers=("customer_unique_id", "nunique"),
        orders=("order_id", "nunique"),
        revenue=("revenue", "sum")
    )
    .sort_values("revenue", ascending=False)
    .reset_index()
)

print("Top 10 States by Revenue:")
print(state_analysis.head(10))
```

```python
print("\nTop 10 States by Customers:")
print(
    state_analysis
    .sort_values("customers", ascending=False)
    .head(10)
)
```

```python
# Payment method analysis

payments = cleaned_data[
    "olist_order_payments_dataset.csv"
].copy()

payment_analysis = (
    payments
    .groupby("payment_type")
    .agg(
        orders=("order_id", "nunique"),
        transactions=("payment_type", "count"),
        total_payment=("payment_value", "sum"),
        average_payment=("payment_value", "mean")
    )
    .sort_values("total_payment", ascending=False)
    .reset_index()
)

print(payment_analysis)
```

```python
# Payment installments distribution

installments = (
    payments
    .groupby("payment_installments")
    .agg(
        transactions=("order_id", "count"),
        total_payment=("payment_value", "sum")
    )
    .sort_values("transactions", ascending=False)
    .reset_index()
)

print("\nPayment Installments:")
print(installments.head(15))
```

```python
# Customer review analysis

reviews = cleaned_data[
    "olist_order_reviews_dataset.csv"
].copy()

print("Total reviews:", len(reviews))

print("\nReview Score Distribution:")
print(
    reviews["review_score"]
    .value_counts()
    .sort_index()
)
```

```python
# Review score percentage

review_percentage = (
    reviews["review_score"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print("\nReview Score Percentage:")
print(review_percentage)
```

```python
print(
    "\nAverage Review Score:",
    round(reviews["review_score"].mean(), 2)
)
```

```python
# Merge reviews with delivery performance

review_delivery = reviews.merge(
    orders[
        [
            "order_id",
            "order_status",
            "delivery_delay_days",
            "delivery_status"
        ]
    ],
    on="order_id",
    how="left"
)

print("Review Delivery Shape:", review_delivery.shape)

print(
    review_delivery[
        [
            "order_id",
            "review_score",
            "delivery_status",
            "delivery_delay_days"
        ]
    ].head(10)
)
```

```python
# Average review score by delivery status

delivery_reviews = (
    review_delivery
    .groupby("delivery_status")
    .agg(
        reviews=("review_score", "count"),
        average_review_score=("review_score", "mean")
    )
    .reset_index()
)

delivery_reviews["average_review_score"] = (
    delivery_reviews["average_review_score"]
    .round(2)
)

print(delivery_reviews)
```

```python
# Recalculate delivery status only for valid delivered orders

valid_delivery = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].notna()) &
    (orders["order_estimated_delivery_date"].notna())
].copy()

valid_delivery["delivery_delay_days"] = (
    valid_delivery["order_delivered_customer_date"]
    - valid_delivery["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)

valid_delivery["delivery_status"] = valid_delivery[
    "delivery_delay_days"
].apply(
    lambda x: "On Time" if x <= 0 else "Late"
)

# Merge only valid delivered orders with reviews

review_delivery_final = reviews.merge(
    valid_delivery[
        [
            "order_id",
            "delivery_delay_days",
            "delivery_status"
        ]
    ],
    on="order_id",
    how="inner"
)

print("Review records with valid delivery data:",
      len(review_delivery_final))

# Average review score by delivery status

delivery_reviews_final = (
    review_delivery_final
    .groupby("delivery_status")
    .agg(
        reviews=("review_score", "count"),
        average_review_score=("review_score", "mean")
    )
    .reset_index()
)

delivery_reviews_final["average_review_score"] = (
    delivery_reviews_final["average_review_score"]
    .round(2)
)

print("\nFinal Delivery vs Review Analysis:")
print(delivery_reviews_final)
```

```python
# Aggregate order items to order level

order_sales = (
    sales_items
    .groupby("order_id")
    .agg(
        product_revenue=("revenue", "sum"),
        freight_value=("freight_value", "sum"),
        items_count=("product_id", "count"),
        unique_products=("product_id", "nunique"),
        unique_sellers=("seller_id", "nunique")
    )
    .reset_index()
)

# Total order value including freight
order_sales["total_order_value"] = (
    order_sales["product_revenue"]
    + order_sales["freight_value"]
)

print("Order-level sales shape:", order_sales.shape)

print("\nSample:")
print(order_sales.head(10))
```

```python
print(
    "Duplicate order IDs:",
    order_sales["order_id"].duplicated().sum()
)
```

```python
# Aggregate payments to order level

order_payments = (
    payments
    .groupby("order_id")
    .agg(
        payment_value=("payment_value", "sum"),
        payment_transactions=("payment_type", "count"),
        payment_methods=("payment_type", "nunique"),
        max_installments=("payment_installments", "max")
    )
    .reset_index()
)

print("Order-level payments shape:", order_payments.shape)

print("\nSample:")
print(order_payments.head(10))
```

```python
print(
    "Duplicate order IDs:",
    order_payments["order_id"].duplicated().sum()
)
```

```python
# Prepare orders for master dataset

master_orders = orders[
    [
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].copy()


# Calculate delivery days

master_orders["delivery_days"] = (
    master_orders["order_delivered_customer_date"]
    - master_orders["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)


# Calculate delivery delay
# Positive = Late
# Zero/negative = On Time

master_orders["delivery_delay_days"] = (
    master_orders["order_delivered_customer_date"]
    - master_orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)


# Delivery status only for delivered orders

master_orders["delivery_status"] = "Not Delivered"

delivered_mask = (
    (master_orders["order_status"] == "delivered") &
    (master_orders["order_delivered_customer_date"].notna()) &
    (master_orders["order_estimated_delivery_date"].notna())
)

master_orders.loc[
    delivered_mask & (master_orders["delivery_delay_days"] <= 0),
    "delivery_status"
] = "On Time"

master_orders.loc[
    delivered_mask & (master_orders["delivery_delay_days"] > 0),
    "delivery_status"
] = "Late"


print("Base orders:", master_orders.shape)

print("\nDelivery Status:")
print(master_orders["delivery_status"].value_counts())

print("\nSample:")
print(
    master_orders[
        [
            "order_id",
            "order_status",
            "delivery_days",
            "delivery_delay_days",
            "delivery_status"
        ]
    ].head(10)
)
```

```python
# Add customer information

customer_info = customers[
    [
        "customer_id",
        "customer_unique_id",
        "customer_city",
        "customer_state"
    ]
].drop_duplicates("customer_id")

master_orders = master_orders.merge(
    customer_info,
    on="customer_id",
    how="left"
)

print("After customer merge:", master_orders.shape)
```

```python
# Add order-level sales

master_orders = master_orders.merge(
    order_sales,
    on="order_id",
    how="left"
)

print("After sales merge:", master_orders.shape)
```

```python
# Add order-level payment information

master_orders = master_orders.merge(
    order_payments,
    on="order_id",
    how="left"
)

print("After payment merge:", master_orders.shape)
```

```python
# Validate master dataset

print("\nFinal Master Dataset:")
print(master_orders.shape)

print(
    "Duplicate order IDs:",
    master_orders["order_id"].duplicated().sum()
)

print("\nSample:")
print(master_orders.head())
```

```python
# Aggregate reviews to order level

order_reviews = (
    reviews
    .groupby("order_id")
    .agg(
        review_score=("review_score", "mean"),
        review_count=("review_id", "count")
    )
    .reset_index()
)

# Round average review score

order_reviews["review_score"] = (
    order_reviews["review_score"].round(2)
)

print("Order-level reviews shape:", order_reviews.shape)

print("\nSample:")
print(order_reviews.head(10))

print(
    "\nDuplicate order IDs:",
    order_reviews["order_id"].duplicated().sum()
)
```

```python
# Add review information to master dataset

master_orders = master_orders.merge(
    order_reviews,
    on="order_id",
    how="left"
)

print("After review merge:", master_orders.shape)

print(
    "Duplicate order IDs:",
    master_orders["order_id"].duplicated().sum()
)
```

```python
# Check review coverage

print("Orders with reviews:")
print(master_orders["review_score"].notna().sum())

print("\nOrders without reviews:")
print(master_orders["review_score"].isna().sum())

print("\nReview score distribution:")
print(
    master_orders["review_score"]
    .value_counts()
    .sort_index()
)
```

```python
# Final missing value check

missing_summary = (
    master_orders.isnull()
    .sum()
    .sort_values(ascending=False)
)

print("Missing Values:")
print(missing_summary[missing_summary > 0])
```

```python
# Final duplicate check

print(
    "Duplicate order IDs:",
    master_orders["order_id"].duplicated().sum()
)
```

```python
# Check payment vs order value reconciliation

reconciliation = master_orders[
    master_orders["payment_value"].notna() &
    master_orders["total_order_value"].notna()
].copy()

reconciliation["payment_difference"] = (
    reconciliation["payment_value"]
    - reconciliation["total_order_value"]
)

print("Orders with both sales and payment data:",
      len(reconciliation))

print("\nTotal payment difference:")
print(
    round(
        reconciliation["payment_difference"].sum(),
        2
    )
)

print("\nOrders where payment != order value:")

mismatch = reconciliation[
    reconciliation["payment_difference"].abs() > 0.01
]

print("Mismatch orders:", len(mismatch))

print("\nMismatch summary:")
print(
    mismatch["payment_difference"]
    .describe()
)
```

```python
# Create clean payment columns

master_orders["payment_value_clean"] = (
    master_orders["payment_value"].fillna(0)
)

master_orders["payment_transactions_clean"] = (
    master_orders["payment_transactions"].fillna(0)
)

master_orders["payment_methods_clean"] = (
    master_orders["payment_methods"].fillna(0)
)

master_orders["max_installments_clean"] = (
    master_orders["max_installments"].fillna(0)
)

print("Payment missing values handled.")
```

```python
# Review availability

master_orders["has_review"] = (
    master_orders["review_score"].notna()
)


# Payment availability

master_orders["has_payment"] = (
    master_orders["payment_value"].notna()
)


# Sales availability

master_orders["has_sales"] = (
    master_orders["total_order_value"].notna()
)


# Delivered order flag

master_orders["is_delivered"] = (
    master_orders["order_status"] == "delivered"
)

print(
    master_orders[
        [
            "order_id",
            "has_review",
            "has_payment",
            "has_sales",
            "is_delivered"
        ]
    ].head(10)
)
```
